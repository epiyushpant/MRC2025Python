```python
import tkinter as tk

root = tk.Tk()
root.title("Dynamic Label")
root.geometry("300x200")

label = tk.Label(root, text="Click the button!", font=("Arial", 14))
label.pack(pady=20)

count = [0]  # use list to allow mutation inside nested function

def update_label():
    count[0] += 1
    label.config(text=f"Clicked {count[0]} times!")  # update text

btn = tk.Button(root, text="Click Me", command=update_label)
btn.pack()

root.mainloop()
```
