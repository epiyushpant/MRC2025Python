 ## Tkinter is Python’s built-in library for creating graphical user interfaces (GUIs).

```python
import tkinter as tk

root = tk.Tk()              # Create the main window
root.title("My First Window")
root.geometry("400x400")    # Set size: width x height

root.mainloop()             # Keep the window open
```

## Button in tkinter 

```python
import tkinter as tk

def say_hello():
    print("Hello, Tkinter!")

root = tk.Tk()
root.title("Button Example")
root.geometry("400x400") 

button = tk.Button(root, text="Click Me", command=say_hello)
button.pack(pady=20)   # Place the button in the window

root.mainloop()
```

### Input in tkinter

```python
import tkinter as tk

root = tk.Tk()
root.title("Entry Example")
root.geometry("400x400") 
label = tk.Label(root, text="Enter your name:")
label.pack()

entry = tk.Entry(root)
entry.pack()

def greet():
    name = entry.get()
    print(f"Hello, {name}!")

button = tk.Button(root, text="Greet", command=greet)
button.pack()

root.mainloop()
```

### Checkbox: allow toggling on or off 

```python
import tkinter as tk

root = tk.Tk()
root.title("Checkbox Example")
root.geometry("400x400") 
var = tk.BooleanVar()

check = tk.Checkbutton(root, text="I agree", variable=var)
check.pack()

def check_status():
    print("Checked:", var.get())

button = tk.Button(root, text="Check Status", command=check_status)
button.pack()

root.mainloop()
```

```python
root.destroy()
```

### Radiobuttons let the user select one option from a group.

```python
import tkinter as tk

root = tk.Tk()
root.title("Radiobutton Example")
root.geometry("400x400") 
choice = tk.StringVar(value="None")

tk.Radiobutton(root, text="Option A", variable=choice, value="A").pack()
tk.Radiobutton(root, text="Option B", variable=choice, value="B").pack()

def show_choice():
    print("Selected:", choice.get())

button = tk.Button(root, text="Submit", command=show_choice)
button.pack()

root.mainloop()
```

## List box to display list of items

```python
import tkinter as tk

root = tk.Tk()
root.title("Listbox Example")
root.geometry("400x400") 

listbox = tk.Listbox(root)
listbox.insert(1, "Apple")
listbox.insert(2, "Banana")
listbox.insert(3, "Cherry")
listbox.pack()

def show_selection():
    selection = listbox.get(listbox.curselection())
    print("You selected:", selection)

button = tk.Button(root, text="Select", command=show_selection)
button.pack()

root.mainloop()
```

### Dropdown in tkinter

```python
import tkinter as tk

root = tk.Tk()
root.title("Dropdown Example")
root.geometry("400x400") 

# Variable to store selected value
selected = tk.StringVar(value="Select an option")

# Create dropdown
options = ["Apple", "Banana", "Cherry", "Date"]
dropdown = tk.OptionMenu(root, selected, *options)
dropdown.pack(pady=10)

def show_choice():
    print("You selected:", selected.get())

button = tk.Button(root, text="Submit", command=show_choice)
button.pack()

root.mainloop()
```

### You can change the background (bg) and foreground (fg) colors of widgets.

```python
import tkinter as tk

root = tk.Tk()
root.title("Colors Example")
root.geometry("400x400") 

label = tk.Label(root, text="Colorful Label", bg="yellow", fg="blue", font=("Arial", 14))
label.pack(pady=10)

button = tk.Button(root, text="Click Me", bg="green", fg="white")
button.pack(pady=10)

root.mainloop()
```

### You can set fonts using the font parameter:

```python
import tkinter as tk

root = tk.Tk()
root.title("Fonts Example")
root.geometry("400x400") 

label = tk.Label(root, text="Big Bold Text", font=("Helvetica", 20, "bold"))
label.pack(pady=10)

button = tk.Button(root, text="Italic Button", font=("Times New Roman", 14, "italic"))
button.pack(pady=10)

root.mainloop()
```

### Checkboxes (Checkbuttons) can also be styled:

```python
import tkinter as tk

root = tk.Tk()
root.title("Checkbox Styling")
root.geometry("400x400")

var = tk.BooleanVar()

check = tk.Checkbutton(root, text="I agree", variable=var, bg="lightgray", fg="red", font=("Arial", 12))
check.pack(pady=10)

root.mainloop()
```

### You can control spacing with padx, pady, and widget dimensions:



```python
import tkinter as tk

root = tk.Tk()
root.title("Padding Example")
root.geometry("400x400") 

button = tk.Button(root, text="Wide Button", width=20, height=2, bg="purple", fg="white")
button.pack(padx=20, pady=20)

root.mainloop()
```

### Styling labels

```python
import tkinter as tk

root = tk.Tk()
root.title("Styled Label")
root.geometry("400x400") 

label = tk.Label(root, text="Styled Label Example",
                 bg="black", fg="white",
                 font=("Helvetica", 18, "bold"),
                 width=30, height=2, anchor="center")
label.pack(pady=10)

root.mainloop()
```

## labels can change text  while program runs 

```python
import tkinter as tk

root = tk.Tk()
root.title("Dynamic Label")
root.geometry("400x400") 

label = tk.Label(root, text="Waiting...", font=("Arial", 14))
label.pack(pady=10)

def update_label():
    label.config(text="Button Clicked!")

button = tk.Button(root, text="Click Me", command=update_label)
button.pack()

root.mainloop()
```

### Labels can display images (PNG, GIF):

```python
import tkinter as tk

root = tk.Tk()
root.title("Image Label")
root.geometry("1000x1000") 
img = tk.PhotoImage(file="pok.png")  # Replace with your image
label = tk.Label(root, image=img)
label.pack()

root.mainloop()
```
