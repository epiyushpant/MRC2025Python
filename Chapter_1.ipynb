## Question 1

<p>Write program to illustrate variables, constants, datatypesand typeconversion. </p>

```python
# Variables
name = "Preeti"
age = 21
height = 5.4
is_student = True

# Constants (by convention, constants are written in uppercase)
PI = 3.14159
GRAVITY = 9.8

# Display data types
print("Name:", name, "| Type:", type(name))
print("Age:", age, "| Type:", type(age))
print("Height:", height, "| Type:", type(height))
print("Is Student:", is_student, "| Type:", type(is_student))
print("PI:", PI, "| Type:", type(PI))

# Type conversion
age_str = str(age)
height_int = int(height)
bool_str = str(is_student)

print("\nType Converted Values:")
print("Age as string:", age_str, "| Type:", type(age_str))
print("Height as int:", height_int, "| Type:", type(height_int))
print("Boolean as string:", bool_str, "| Type:", type(bool_str))
```

## question 2

<p>Program to demonstrate different types of operators and perform calculations.</p>

```python
# Arithmetic Operators
a = 10
b = 3

print("Arithmetic Operators:")
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a % b =", a % b)
print("a ** b =", a ** b)
print("a // b =", a // b)

# Assignment Operators
x = 5
x += 3
print("\nAssignment Operator: x += 3 →", x)

# Comparison Operators
print("\nComparison Operators:")
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)

# Logical Operators
print("\nLogical Operators:")
print("a > 5 and b < 5:", a > 5 and b < 5)
print("a > 5 or b > 5:", a > 5 or b > 5)
print("not(a > 5):", not(a > 5))

# Bitwise Operators
print("\nBitwise Operators:")
print("a & b:", a & b)
print("a | b:", a | b)
print("a ^ b:", a ^ b)
```
