## What Are Strings?

A string is just a bunch of letters, numbers, or symbols put together, like "Hello" or "Ram123"

We'll use a *class* called `StringOperations` to organize our strings.

## The StringOperations Class

```python
class StringOperations:
    def __init__(self):
        try:
            input_string = input("Type a word (like 'ram' or 'shyam'): ")
            if not input_string.isalpha():
                raise TypeError("Oops! Please enter a word with only letters.")
            self.text = input_string
        except Exception as e:
            print(f"Something went wrong: {str(e)}")
            self.text = ""

your_input = StringOperations()
print("Your word is:", your_input.text)
```

**What's Happening?**

- We made a `StringOperations` class, your_input to hold our word.
- It asks you to type a word (like "input_string").
- The `try` part is like trying to catch a  cricket ball—if it works wicket, If not, `except` catches the mistake and tells us what went wrong in catch.

## QN 1: How Long Is Your Word?

Let's count how many letters are in your word.

```python
class StringOperations:
    def __init__(self):
        try:
            input_string = input("Type a fun word (like 'gpoi' or 'nima'): ")
            if not input_string.isalpha():
                raise TypeError("We need a word, not something else.")
            self.text = input_string
        except Exception as e:
            print(f"Something went wrong: {str(e)}")
            self.text = ""

    def get_length(self):
        try:
            return len(self.text)
        except Exception as e:
            return f"Something went wrong: {str(e)}"

your_text = StringOperations()
print("Your word is:", your_text.text)
print("It has", your_text.get_length(), "letters!")
```

**What's Happening?**

- The `get_length` trick counts the letters in your word (like "ram" has 3 letters).
- If something goes wrong (like if the word is missing), our safety net (`try` and `except`) catches it and says, "Oops!"

## QN 2: Glue Two Words Together

```python
class StringOperations:
    def __init__(self):
        try:
            self.text = input("Type a  word just like above: ")
            if not self.text.isalpha():
                raise TypeError("Please enter a word with only alphabets")
        except TypeError as e:
            print(f"Meow {e}")
            self.text = ""

    def concatenate(self, other_string):
        if not isinstance(other_string, str):
            return "We can only glue a word, not anything else"
        if not self.text:
            return "No valid word to glue with!"
        return self.text + other_string

your_input = StringOperations()
print("Your word is:", your_input.text)
print("After gluing ' World':", your_input.concatenate(" World"))
```

**What's Happening?**

- The `concatenate` trick glues your word with another word.

## QN 3: Cut a Piece of Your Word

```python
class StringOperations:
    def __init__(self):
        try:
            input_string = input("Type a word: ")
            if not isinstance(input_string, str):
                raise TypeError("Oops! We need a string not something else.")
            self.text = input_string
        except Exception as e:
            print(f"Something went wrong: {str(e)}")
            self.text = ""

    def slice_string(self, start, end):
        try:
            return self.text[start:end]
        except (IndexError, TypeError) as e:
            return f"Something went wrong: {str(e)}"

user_input = StringOperations()
print("Your word is:", user_input.text)
print("First 3 letters:", user_input.slice_string(0, 3))
```

## QN 4: Make Our Word CAPITAL!

Let's make your word capital by turning it all to big letters

```python
class StringOperations:
    def __init__(self):
        try:
            input_string = input("Type any word : ")
            if not isinstance(input_string, str):
                raise TypeError("Oops! We need stringonly.")
            self.text = input_string
        except Exception as e:
            print(f"Something went wrong: {str(e)}")
            self.text = ""

    def to_uppercase(self):
        try:
            return self.text.upper()
        except AttributeError as e:
            return f"Oops! Something went wrong: {str(e)}"
your_input = StringOperations()
print("Your word is:", your_input.text)
print("Your word shouting:", your_input.to_uppercase())
```

**What's Happening?**

- The `to_uppercase` trick turns your word into all big letters.

## QN 5: Break Word Apart

```python
class StringOperations:
    def __init__(self):
        try:
            self.text = input("Type a sequence of words: ")
            if not self.text:
                raise ValueError("ERROR")
        except ValueError as e:
            print(f"Uh-oh! {e}")
            self.text = ""

    def split_string(self):
        try:
            delimiter = input("Type the delimiter (e.g., ' ' or ','): ")
            if not isinstance(delimiter, str):
                raise TypeError("Oops! Delimiter must be a string.")
            return self.text.split(delimiter)
        except TypeError as e:
            print(f"Uh-oh! {e}")
            return [self.text]  

your_data = StringOperations()
print("Your word or phrase is:", your_data.text)
print("Split result:", your_data.split_string())
```

**What's Happening?**

- The `is_alphanumeric` trick checks if your word has only letters and numbers (no symbols like "!").
- It says `True` if it’s okay, or `False` if it has other stuff. Our safety net keeps things safe!
