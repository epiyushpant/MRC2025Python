```python
total = 0
count = 0

while True:
    num = int(input("Enter a number (negative to stop): "))

    if num < 0:  # Stop when a negative number is entered
        print("Exiting program...")
        break

    if num % 5 == 0:  # Skip multiples of 5
        print("Skipping this number.")
        continue

    total += num  # Add valid numbers
    count += 1  # Count valid numbers

print(f"You entered {count} valid numbers. Sum: {total}")
```
