**1. Write a program to check if a number is positive, negative or zero using if**

```python
num = int(input("Enter a number: "))
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

**2. Use match to create a simple menu-based calculator**

```python
op = input("Choose operation (add/sub): ")
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
match op:
    case "add": print(a + b)
    case "sub": print(a - b)
    
```

**3. Use break to stop input when the user enters a specific keyword.**

```python
while True:
    word = input("Enter something (or 'stop' to break): ")
    if word == "stop":
        break
```

**4. Use continue to skip printing even numbers in a loop.**


```python
for i in range(1, 11):  # Loop from 1 to 10
    if i % 2 == 0:  # Check if the number is even
        continue  # Skip the current iteration
    print(i)  # Print odd numbers
```

**5. Find the factorial of a number using both for and while loops.**

```python
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n+1):
    fact *= i
print("Factorial:", fact)
```

**6.Print the first 10 Fibonacci numbers.**

```python
a, b = 0, 1
for _ in range(10):
    print(a, end=' ')
    a, b = b, a + b

# “I’m looping 10 times, but I don’t care what the index is.”
```

**7. Sum of all numbers from 1 to N entered by the user.**

```python
N = int(input("Enter N: "))
sum=0;
for i in range(1, N+1):
    sum+=i 
print("Sum:", sum)
```

**8. Keep asking the user to input a valid age (between 1 to 100) using a loop.**


```python
while True:
    age = int(input("Enter valid age (1-100): "))
    if 1 <= age <= 100:
        break
print("Valid age:", age)
```

 **10.Print a number pyramid pattern.**

```python
for i in range(1, 6):
    for j in range(1, i+1):
        print(j, end=' ')
    print()
```

**11. Create a menu-driven program using infinite loop and break to allow user to choose options like add, subtract, exit etc.**

```python
while True:
    print("1. Add\n2. Subtract\n3. Exit")
    choice = input("Enter choice: ")
    if choice == "3":
        break
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    if choice == "1":
        print("Result:", a + b)
    elif choice == "2":
        print("Result:", a - b)
```
