```python
num = int(input("Enter a number: "))
if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

```python
op = input("Choose operation (add/sub): ")
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
match op:
    case "add": print(a + b)
    case "sub": print(a - b)
```

```python
while True:
    word = input("Enter something (or 'stop' to break): ")
    if word == "stop":
        break
```

```python
n = int(input("Enter a number: "))
fact = 1
for i in range(1, n+1):
    fact *= i
print("Factorial:", fact)
```

```python
a, b = 0, 1
for _ in range(10):
    print(a, end=' ')
    a, b = b, a + b
```

```python
N = int(input("Enter N: "))
total = sum(range(1, N+1))
print("Sum:", total)
```

```python
while True:
    age = int(input("Enter valid age (1-100): "))
    if 1 <= age <= 100:
        break
print("Valid age:", age)
```

```python
for i in range(1, 6):
    print("*" * i)
```

```python
for i in range(1, 6):
    for j in range(1, i+1):
        print(j, end=' ')
    print()
```

```python
while True:
    print("1. Add\n2. Subtract\n3. Exit")
    choice = input("Enter choice: ")
    if choice == "3":
        break
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    if choice == "1":
        print("Result:", a + b)
    elif choice == "2":
        print("Result:", a - b)
```
