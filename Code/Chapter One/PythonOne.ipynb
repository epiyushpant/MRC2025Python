# 🔄 Python Type Conversion
## 1. Implicit Type Conversion (Type Casting by Python)
Done automatically by Python when it’s safe.
Converting from small data type to large , no loss of data 


````python
x = 10      # int
y = 2.5     # float
z = x + y   # Python converts int → float
print(z)    # 12.5 (float)
````

 Python promotes int → float → complex as needed.

## 2. Explicit Type Conversion (Type Casting by User)
Done manually using built-in functions.

Common functions: int(), float(), str(), list(), tuple(), set(), dict(), complex(), bool().

````python
# int → float
x = float(5)        # 5.0

# float → int
y = int(2.8)        # 2 (truncates decimal)

# number → string
z = str(10)         # "10"

# string → int
a = int("25")       # 25

# list → tuple
b = tuple([1, 2, 3])   # (1, 2, 3)

# tuple → set
c = set((1, 2, 2, 3))  # {1, 2, 3}

# list → dict (must be key-value pairs)
d = dict([("name", "John"), ("age", 30)])  # {"name": "John", "age": 30}
````

````python
print(x)
print(y)
print(z)
print(a)
print(b)
print(c)
print(d)
````

| Function   | Converts To | Example |
|------------|-------------|---------|
| int()      | Integer     | int("5") → 5 |
| float()    | Float       | float("3.14") → 3.14 |
| str()      | String      | str(25) → "25" |
| list()     | List        | list((1,2,3)) → [1,2,3] |
| tuple()    | Tuple       | tuple([1,2,3]) → (1,2,3) |
| set()      | Set         | set([1,2,2,3]) → {1,2,3} |
| frozenset()| Frozenset   | frozenset([1,2,3]) → frozenset({1,2,3}) |
| dict()     | Dictionary  | dict([("a",1),("b",2)]) → {"a":1,"b":2} |
| complex()  | Complex     | complex(5) → (5+0j) |
| bool()     | Boolean     | bool(0) → False |


- Implicit conversion: Python does it automatically when safe.
- Explicit conversion: You control it using functions.
- Always ensure the data is compatible (e.g., int("abc") will throw an error).

## 1.5 Reading input from user 
In Python, you can read input from the keyboard using the built‑in input() function. It always returns the data as a string, so you often need to convert it to the type you want.

````python
# Reading from user
name = input("Enter your name: ")
print("Hello,", name)
````

**Converting Input to Other Types**
Since input() returns a string, you can wrap it with type conversion functions:

````python
age = int(input("Enter your age: "))       # converts to int
height = float(input("Enter your height: ")) # converts to float
is_student = bool(input("Are you a student? ")) # converts to bool (non-empty string → True)
````

````python
print(age,height,is_student)
````

````python
#Reading multiple values
x, y = input("Enter two numbers: ").split()
x = int(x)
y = int(y)
print("Sum:", x + y)
````

````python
#Reading List of values
fruits = input("Enter fruits separated by space: ").split()
print(fruits)   # ['apple', 'banana', 'cherry']
````

````python
#reading number in a list
numbers = list(map(int, input("Enter numbers: ").split()))
print(numbers)  # [10, 20, 30]
````

## 🎨 Formatted Output with f-strings
Starting from Python 3.6, f-strings (formatted string literals) make it easy to embed expressions inside strings.

````python
name = "Alice"
age = 25
print(f"My name is {name} and I am {age} years old.")
````

Anything inside {} is evaluated as Python code.

````python
#expression inside f string
x = 10
y = 5
print(f"Sum of {x} and {y} is {x + y}")
# Output: Sum of 10 and 5 is 15
````

````python
#Formatting numbers 
pi = 3.141592653589793
print(f"Pi rounded to 2 decimals: {pi:.2f}")
# Output: Pi rounded to 2 decimals: 3.14
````

#Aligning text 
name = "Bob"
print(f"{name:<10} left aligned")   # Bob       left aligned
print(f"{name:^10} centered")       #    Bob    centered
print(f"{name:>10} right aligned")  #       Bob right aligned


````python
#using function inside f string 
name = "alice"
print(f"Capitalized name: {name.capitalize()}")
# Output: Capitalized name: Alice
````
