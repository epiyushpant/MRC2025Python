```python
# 1. Print maximum of two input numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

#print("Maximum number:", max(num1, num2))

if num1 > num2:
    max_num = num1
else:
    max_num = num2

print("Maximum number:", max_num)
```

```python
# 2. Print area of rectangle
length = float(input("Enter length of the rectangle: "))
width = float(input("Enter width of the rectangle: "))
print("Area of Rectangle:", length * width)

```

```python
# 3. Print Simple Interest
principal = float(input("Enter principal amount: "))
rate = float(input("Enter rate of interest (in %): "))
time = float(input("Enter time (in years): "))
simple_interest = (principal * rate * time) / 100
print("Simple Interest:", simple_interest)

```

```python
# 4. Check If Number Is Prime or Not
number = int(input("Enter a number to check if it's prime: "))
is_prime = True if number > 1 else False
for i in range(2, number/2):
    if number % i == 0:
        is_prime = False
        break
print(f"{number} is Prime" if is_prime else f"{number} is Not Prime")

```

```python
# 5. Sum of Numbers 1 to N
N = int(input("Enter a number N to find sum from 1 to N: "))
# sum_N = N * (N + 1) // 2
# print(f"Sum of numbers from 1 to {N}: {sum_N}")

fact = 1

for i in range(1, N + 1):
    fact = fact * i  # Multiplying fact by i in each iteration

print(f"Factorial of {N} is {fact}")

```

```python
#6.Factorial 
N = int(input("Enter a number: "))
fact = 1

for i in range(1, N + 1):
    fact *= i  # Multiply fact by i in each iteration

print(f"Factorial of {N} is {fact}")
```

```python
#Using recursion 
# 6. Factorial of N
def factorial(n):
    return 1 if n == 0 else n * factorial(n - 1)

N_fact = int(input("Enter a number to find its factorial: "))
print(f"Factorial of {N_fact}: {factorial(N_fact)}")
```
