## 1. Program to Establish Connection with Database and Create or Delete Database and Table

This program uses SQLite to create a database file (`school.db`), create a table (`students`), and demonstrate deleting a table. SQLite stores the database in a file, so no external server is needed.

```python
import sqlite3

def manage_database():
    try:
        # Connect to database (creates school.db if it doesn't exist)
        conn = sqlite3.connect('school.db')
        cursor = conn.cursor()
        print("Connected to database.")

        # Create table
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS students (
                id INTEGER PRIMARY KEY,
                name TEXT NOT NULL,
                age INTEGER
            )
        ''')
        print("Table 'students' created or already exists.")

        # Verify table creation
        cursor.execute("SELECT name FROM sqlite_master WHERE type='table' AND name='students'")
        if cursor.fetchone():
            print("Table 'students' exists.")

        # Delete table (for demonstration)
        cursor.execute("DROP TABLE IF EXISTS students")
        print("Table 'students' deleted.")

        # Commit changes and close connection
        conn.commit()
        conn.close()
        print("Database connection closed.")

    except sqlite3.Error as e:
        print(f"Database error: {e}")

manage_database()
```

## 2. Program to Store Data in Database and Manipulate the Data

This program creates a `students` table, inserts data, updates a record, and deletes a record.

```python
import sqlite3

def manipulate_data():
    try:
        # Connect to database
        conn = sqlite3.connect('school.db')
        cursor = conn.cursor()

        # Create table
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS students (
                id INTEGER PRIMARY KEY,
                name TEXT NOT NULL,
                age INTEGER
            )
        ''')

        # Insert data
        cursor.executemany('INSERT INTO students (name, age) VALUES (?, ?)', 
                           [('Nima', 20), ('Gopi', 22), ('Pradeep', 19)])
        print("Data inserted.")

        # Update data
        cursor.execute("UPDATE students SET age = ? WHERE name = ?", (21, 'Gopi'))
        print("Updated Gopi's age.")

        # Delete data
        cursor.execute("DELETE FROM students WHERE name = ?", ('Pradeep',))
        print("Deleted Pradeep's record.")

        # Display data
        cursor.execute("SELECT * FROM students")
        rows = cursor.fetchall()
        print("Current data in students table:")
        for row in rows:
            print(row)

        # Commit and close
        conn.commit()
        conn.close()

    except sqlite3.Error as e:
        print(f"Database error: {e}")

manipulate_data()
```

## 3. Program to Perform CRUD Operations in Database

This program implements CRUD operations (Create, Read, Update, Delete) on the `students` table in an interactive manner.

```python
import sqlite3

def crud_operations():
    try:
        # Connect to database
        conn = sqlite3.connect('school.db')
        cursor = conn.cursor()

        # Create table
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS students (
                id INTEGER PRIMARY KEY,
                name TEXT NOT NULL,
                age INTEGER
            )
        ''')

        # CRUD: Create
        cursor.execute("INSERT INTO students (name, age) VALUES (?, ?)", ('Chetan', 23))
        print("Created record for Chetan.")

        # CRUD: Read
        cursor.execute("SELECT * FROM students WHERE name = ?", ('Chetan',))
        print("Read record:", cursor.fetchone())

        # CRUD: Update
        cursor.execute("UPDATE students SET age = ? WHERE name = ?", (24, 'Chetan'))
        print("Updated Chetan's age.")

        # CRUD: Read after update
        cursor.execute("SELECT * FROM students WHERE name = ?", ('Chetan',))
        print("Updated record:", cursor.fetchone())

        # CRUD: Delete
        cursor.execute("DELETE FROM students WHERE name = ?", ('Chetan',))
        print("Deleted Chetan's record.")

        # Verify deletion
        cursor.execute("SELECT * FROM students WHERE name = ?", ('Chetan',))
        if not cursor.fetchone():
            print("Chetan's record no longer exists.")

        # Commit and close
        conn.commit()
        conn.close()

    except sqlite3.Error as e:
        print(f"Database error: {e}")

crud_operations()
```

## 4. Program to Create Simple GUI with Widgets: Label, Text Entry, Radio Buttons, Check Buttons

This program creates a Tkinter GUI with a label, text entry, radio buttons (for selecting gender), and check buttons (for selecting hobbies).

```python
import tkinter as tk
from tkinter import messagebox

def submit_form():
    name = entry_name.get()
    gender = gender_var.get()
    hobbies = [h for h, var in hobby_vars.items() if var.get()]
    messagebox.showinfo("Submission", f"Name: {name}\nGender: {gender}\nHobbies: {', '.join(hobbies)}")

# Create main window
root = tk.Tk()
root.title("Simple GUI Form")
root.geometry("300x300")

# Label
tk.Label(root, text="Enter Your Name:").pack(pady=5)

# Text Entry
entry_name = tk.Entry(root)
entry_name.pack(pady=5)

# Radio Buttons (Gender)
gender_var = tk.StringVar(value="Male")
tk.Label(root, text="Select Gender:").pack(pady=5)
tk.Radiobutton(root, text="Male", variable=gender_var, value="Male").pack()
tk.Radiobutton(root, text="Female", variable=gender_var, value="Female").pack()

# Check Buttons (Hobbies)
tk.Label(root, text="Select Hobbies:").pack(pady=5)
hobby_vars = {"Reading": tk.BooleanVar(), "Gaming": tk.BooleanVar(), "Sports": tk.BooleanVar()}
for hobby, var in hobby_vars.items():
    tk.Checkbutton(root, text=hobby, variable=var).pack()

# Submit Button
tk.Button(root, text="Submit", command=submit_form).pack(pady=10)

# Run the application
root.mainloop()
```

## 5. Program to Organize Different Widgets with Frame to Create Attractive Designs

This program uses Tkinter frames to organize widgets (labels, entries, buttons) into sections for a user registration form with a cleaner layout.

```python
import tkinter as tk
from tkinter import messagebox

def register():
    username = entry_username.get()
    password = entry_password.get()
    email = entry_email.get()
    messagebox.showinfo("Registration", f"Registered: {username}\nEmail: {email}")

# Create main window
root = tk.Tk()
root.title("Registration Form")
root.geometry("400x300")

# Frame for personal info
personal_frame = tk.Frame(root, borderwidth=2, relief="groove", padx=10, pady=10)
personal_frame.pack(pady=10, fill="x")

tk.Label(personal_frame, text="Personal Information", font=("Arial", 12, "bold")).pack()
tk.Label(personal_frame, text="Username:").pack()
entry_username = tk.Entry(personal_frame)
entry_username.pack(pady=5)

# Frame for account info
account_frame = tk.Frame(root, borderwidth=2, relief="groove", padx=10, pady=10)
account_frame.pack(pady=10, fill="x")

tk.Label(account_frame, text="Account Information", font=("Arial", 12, "bold")).pack()
tk.Label(account_frame, text="Password:").pack()
entry_password = tk.Entry(account_frame, show="*")
entry_password.pack(pady=5)
tk.Label(account_frame, text="Email:").pack()
entry_email = tk.Entry(account_frame)
entry_email.pack(pady=5)

# Frame for buttons
button_frame = tk.Frame(root, pady=10)
button_frame.pack()
tk.Button(button_frame, text="Register", command=register).pack(side="left", padx=5)
tk.Button(button_frame, text="Clear", command=lambda: [entry_username.delete(0, tk.END), entry_password.delete(0, tk.END), entry_email.delete(0, tk.END)]).pack(side="left", padx=5)

# Run the application
root.mainloop()
```

## 6. Program to Draw Different Geometrical Shapes Using Canvas Widget

This program uses Tkinter's Canvas widget to draw a rectangle, circle, and triangle.

```python
import tkinter as tk

# Create main window
root = tk.Tk()
root.title("Geometrical Shapes")
root.geometry("400x400")

# Create canvas
canvas = tk.Canvas(root, width=350, height=350, bg="white")
canvas.pack(pady=20)

# Draw rectangle
canvas.create_rectangle(50, 50, 150, 100, fill="blue", outline="black")

# Draw circle (oval with equal width/height)
canvas.create_oval(200, 50, 300, 150, fill="red", outline="black")

# Draw triangle (polygon with 3 points)
canvas.create_polygon(100, 200, 50, 300, 150, 300, fill="green", outline="black")

# Labels for shapes
canvas.create_text(100, 125, text="Rectangle", font=("Arial", 10))
canvas.create_text(250, 175, text="Circle", font=("Arial", 10))
canvas.create_text(100, 325, text="Triangle", font=("Arial", 10))

# Run the application
root.mainloop()
```
