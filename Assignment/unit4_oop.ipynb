## 1. Program to Elaborate Object-Oriented Concepts with Simple Examples

This program demonstrates basic OOP concepts:
- **Class**: Blueprint for creating objects (e.g., `Dog`).
- **Object**: Instance of a class (e.g., `my_dog`).
- **Attributes**: Data stored in objects (e.g., `name`, `age`).
- **Methods**: Functions defined in classes (e.g., `bark()`).

We create a simple `Dog` class to illustrate these.

```python
# Define a class
class Dog:
    # Class attribute (shared by all instances)
    species = "Canine"
    
    # Instance method
    def bark(self):
        return "Woof!"
    
    # Another method
    def describe(self, name, age):
        return f"{name} is a {self.species} and is {age} years old."

# Create objects (instances)
dog1 = Dog()
dog2 = Dog()

# Access class attribute
print("Species:", dog1.species)

# Call methods
print("Dog1 says:", dog1.bark())
print(dog1.describe("Buddy", 5))

# Add instance attributes
dog1.name = "Buddy"
dog1.age = 5
dog2.name = "Max"
dog2.age = 3

print(f"{dog1.name} is {dog1.age} years old.")
print(f"{dog2.name} is {dog2.age} years old.")
```

## 2. Program to Use __init__ Method to Initialize Objects

The `__init__` method (constructor) is called when an object is created. It initializes instance attributes. This program shows how to use `__init__` in a `Person` class.

```python
class Person:
    # Constructor (__init__ method)
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute
    
    # Method to display info
    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."

# Create objects with initialization
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

# Access attributes and call method
print(person1.greet())
print(person2.greet())

# Modify attribute
person1.age = 31
print(f"Updated: {person1.greet()}")
```

## 3. Program to Apply Different Types of Inheritance

Inheritance allows a class to inherit attributes and methods from another class. This program demonstrates:
- **Single Inheritance**: One base class.
- **Multiple Inheritance**: Multiple base classes.
- **Multilevel Inheritance**: Chain of inheritance.

```python
# Single Inheritance
class Animal:  # Base class
    def eat(self):
        return "Eating..."

class Cat(Animal):  # Derived class
    def meow(self):
        return "Meow!"

cat = Cat()
print("Single Inheritance:")
print(cat.eat())  # Inherited method
print(cat.meow()) # Derived method

# Multiple Inheritance
class Flyer:
    def fly(self):
        return "Flying..."

class Swimmer:
    def swim(self):
        return "Swimming..."

class Duck(Flyer, Swimmer):  # Inherits from both
    def quack(self):
        return "Quack!"

duck = Duck()
print("\nMultiple Inheritance:")
print(duck.fly())
print(duck.swim())
print(duck.quack())

# Multilevel Inheritance
class Mammal(Animal):  # Inherits from Animal
    def walk(self):
        return "Walking..."

class Human(Mammal):  # Inherits from Mammal
    def speak(self):
        return "Speaking..."

human = Human()
print("\nMultilevel Inheritance:")
print(human.eat())  # From Animal
print(human.walk()) # From Mammal
print(human.speak())# From Human
```

## 4. Program to Elaborate Polymorphism and Data Hiding Concepts

- **Polymorphism**: Allows methods to do different things based on the object (method overriding).
- **Data Hiding (Encapsulation)**: Restricts access to attributes/methods using single (_) or double (__) underscores.

This program shows polymorphism via method overriding and data hiding with private attributes.

```python
# Polymorphism (Method Overriding)
class Shape:
    def area(self):
        return "Area of shape"

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    def area(self):  # Overrides base method
        return self.length * self.width

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):  # Overrides base method
        return 3.14 * self.radius * self.radius

# Polymorphism in action
shapes = [Rectangle(4, 5), Circle(3)]
print("Polymorphism:")
for shape in shapes:
    print("Area:", shape.area())  # Calls different implementations

# Data Hiding (Encapsulation)
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute (name mangling)
    
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
    
    def get_balance(self):  # Public method to access private attribute
        return self.__balance

account = BankAccount(1000)
account.deposit(500)
print("\nData Hiding:")
print("Balance:", account.get_balance())  # Access via method
# print(account.__balance)  # This would raise an error (hidden)
```
