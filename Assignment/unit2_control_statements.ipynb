## 1. Program to Apply if, match, break, and continue Statements for Decision Making

This program demonstrates:
- `if-elif-else` for checking conditions.
- `match` (structural pattern matching, Python 3.10+).
- `break` to exit a loop early.
- `continue` to skip iterations.

The program simulates a number guessing game where the user guesses a number between 1 and 10. It uses a loop with continue (to skip invalid guesses) and break (to exit on correct guess or quit). Inside, it uses if for range checks and match to provide feedback based on the guess.

```python
# Secret number to guess
secret_number = 7

# Loop for guesses
attempts = 0
while attempts < 5:
    guess = input("Guess a number between 1 and 10 (or 'q' to quit): ")
    
    if guess == 'q':
        print("Quitting the game.")
        break  # Exit the loop
    
    if not guess.isdigit():
        print("Invalid input. Please enter a number.")
        continue  # Skip to next iteration
    
    guess = int(guess)
    attempts += 1
    
    # Use if-elif-else for range check
    if guess < 1 or guess > 10:
        print("Guess out of range. Try again.")
        continue
    elif guess == secret_number:
        print("Correct! You guessed it.")
        break
    else:
        # Use match for feedback
        match guess:
            case _ if guess < secret_number:
                print("Too low!")
            case _ if guess > secret_number:
                print("Too high!")
            case _:
                print("Unexpected error.")

if attempts == 5:
    print("Out of attempts. The secret was", secret_number)
```

## 2. Program to Utilize Different Loop Statements to Solve Meaningful Problems

This section includes two examples:
- A `for` loop to calculate the factorial of a number (meaningful for math/combinatorics).
- A `while` loop for a simple banking system simulation (meaningful for transaction processing).

```python
# For loop: Calculate factorial
def calculate_factorial(n):
    if n < 0:
        return "Factorial not defined for negative numbers."
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    return factorial

print("Factorial of 5:", calculate_factorial(5))  # Output: 120

# While loop: Simple banking withdrawal simulation
balance = 1000
while balance > 0:
    withdraw = float(input("Enter amount to withdraw (0 to exit): "))
    if withdraw == 0:
        break
    if withdraw > balance:
        print("Insufficient funds. Balance:", balance)
        continue
    balance -= withdraw
    print("Withdrawal successful. New balance:", balance)

print("Final balance:", balance)
```

## 3. Program to Demonstrate Input Validation Using Loop

This program uses a `while` loop to validate user input, ensuring the entered age is a positive integer between 18 and 100.

```python
# Input validation loop
valid = False
while not valid:
    age_input = input("Enter your age (must be between 18 and 100): ")
    if age_input.isdigit():
        age = int(age_input)
        if 18 <= age <= 100:
            valid = True
            print("Valid age entered:", age)
        else:
            print("Age out of range. Try again.")
    else:
        print("Invalid input. Must be a number. Try again.")
```

## 4. Program to Create Different Patterns Using Nested Loops

This program uses nested `for` loops to print two patterns:
- A right-angled triangle of stars.
- A square of numbers.

```python
# Pattern 1: Right-angled triangle
rows = 5
for i in range(1, rows + 1):
    for j in range(i):
        print("*", end=" ")
    print()
    
print("\n")

# Pattern 2: Square of numbers
size = 4
for i in range(1, size + 1):
    for j in range(1, size + 1):
        print(j, end=" ")
    print()
```

## 5. Program to Make Use of Infinite Loop

This program uses an infinite `while True` loop to create a simple menu-driven calculator. It runs indefinitely until the user chooses to exit (using `break`).

```python
# Infinite loop for a menu-driven calculator
while True:
    print("\nCalculator Menu:")
    print("1. Add")
    print("2. Subtract")
    print("3. Exit")
    
    choice = input("Enter choice (1/2/3): ")
    
    if choice == '3':
        print("Exiting calculator.")
        break
    
    if choice not in ['1', '2']:
        print("Invalid choice. Try again.")
        continue
    
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    
    if choice == '1':
        print("Result:", num1 + num2)
    elif choice == '2':
        print("Result:", num1 - num2)
```
