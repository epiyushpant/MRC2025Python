## 1. Program to Create List, Add Elements, Remove Elements, and Display List Items

This program demonstrates creating a list, adding elements (using append and extend), removing elements (using remove and pop), and displaying the list.

```python
# Create a list
fruits = ["apple", "banana", "orange"]
print("Initial list:", fruits)

# Add elements
fruits.append("grape")  # Add single element
fruits.extend(["mango", "kiwi"])  # Add multiple elements
print("After adding elements:", fruits)

# Remove elements
fruits.remove("banana")  # Remove by value
popped_item = fruits.pop()  # Remove last element
print(f"After removing 'banana' and popping '{popped_item}':", fruits)

# Display list items
print("Final list items:")
for index, item in enumerate(fruits):
    print(f"Index {index}: {item}")
```

## 2. Program to Use List Slicing to Display Elements

This program demonstrates list slicing to extract and display specific portions of a list.

```python
# Create a list
numbers = [10, 20, 30, 40, 50, 60, 70, 80]
print("Original list:", numbers)

# Slicing examples
print("First three elements:", numbers[:3])  # Slice from start to index 2
print("Last three elements:", numbers[-3:])  # Slice last three elements
print("Elements from index 2 to 5:", numbers[2:6])  # Slice from index 2 to 5
print("Every second element:", numbers[::2])  # Slice with step of 2
print("Reversed list:", numbers[::-1])  # Reverse the list
```

## 3. Program to Elaborate Different List Methods

This program demonstrates common list methods like append, insert, remove, pop, clear, index, count, sort, and reverse.

```python
# Create a list
items = [3, 1, 4, 1, 5]
print("Initial list:", items)

# append: Add element to end
items.append(9)
print("After append(9):", items)

# insert: Add element at specific index
items.insert(2, 7)
print("After insert(2, 7):", items)

# remove: Remove first occurrence of value
items.remove(1)
print("After remove(1):", items)

# pop: Remove and return element at index
popped = items.pop(3)
print(f"After pop(3) (popped {popped}):", items)

# index: Find index of first occurrence
print("Index of 5:", items.index(5))

# count: Count occurrences of value
print("Count of 1:", items.count(1))

# sort: Sort the list
items.sort()
print("After sort():", items)

# reverse: Reverse the list
items.reverse()
print("After reverse():", items)

# clear: Remove all elements
items.clear()
print("After clear():", items)
```

## 4. Program to Apply List Comprehension

This program uses list comprehension to create lists based on conditions, such as squares of numbers, even numbers, and filtered strings.

```python
# List comprehension examples
# Squares of numbers from 1 to 10
squares = [x**2 for x in range(1, 11)]
print("Squares of 1 to 10:", squares)

# Even numbers from 1 to 20
evens = [x for x in range(1, 21) if x % 2 == 0]
print("Even numbers from 1 to 20:", evens)

# Filter strings longer than 3 characters
words = ["cat", "elephant", "dog", "rhinoceros", "bird"]
long_words = [word for word in words if len(word) > 3]
print("Words longer than 3 characters:", long_words)
```

## 5. Program to Illustrate Two-Dimensional List

This program creates a 2D list (matrix), accesses elements, and performs a simple operation (transpose).

```python
# Create a 2D list (3x3 matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Original matrix:")
for row in matrix:
    print(row)

# Access specific element
print("Element at [1][2]:", matrix[1][2])  # Row 1, Column 2

# Transpose the matrix using list comprehension
transpose = [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
print("Transposed matrix:")
for row in transpose:
    print(row)
```

## 6. Program to Create Tuple, Add Elements, Remove Elements, and Display Tuple Items

Tuples are immutable, so adding/removing elements requires creating new tuples. This program demonstrates tuple creation, conversion to list for modifications, and displaying items.

```python
# Create a tuple
colors = ("red", "blue", "green")
print("Initial tuple:", colors)

# Add elements (convert to list, modify, convert back)
temp_list = list(colors)
temp_list.append("yellow")
colors = tuple(temp_list)
print("After adding 'yellow':", colors)

# Remove elements (convert to list, remove, convert back)
temp_list.remove("blue")
colors = tuple(temp_list)
print("After removing 'blue':", colors)

# Display tuple items
print("Tuple items:")
for index, item in enumerate(colors):
    print(f"Index {index}: {item}")
```

## 7. Program to Create Dictionary, Add Elements, Remove Elements, and Display Dictionary Items

This program demonstrates dictionary creation, adding key-value pairs, removing items, and displaying contents.

```python
# Create a dictionary
student = {"name": "Alice", "age": 20, "major": "Physics"}
print("Initial dictionary:", student)

# Add elements
student["grade"] = "A"
student.update({"id": 12345, "year": 2})
print("After adding elements:", student)

# Remove elements
del student["age"]
popped_item = student.pop("major")
print(f"After removing 'age' and popping 'major' ({popped_item}):", student)

# Display dictionary items
print("Dictionary items:")
for key, value in student.items():
    print(f"{key}: {value}")
```

## 8. Program to Create Set, Add Elements, Remove Elements, and Display Set Items

This program demonstrates set creation, adding/removing elements, and displaying items.

```python
# Create a set
numbers = {1, 2, 3, 4}
print("Initial set:", numbers)

# Add elements
numbers.add(5)
numbers.update([6, 7])
print("After adding elements:", numbers)

# Remove elements
numbers.remove(2)  # Raises error if not found
numbers.discard(3)  # No error if not found
print("After removing 2 and 3:", numbers)

# Display set items
print("Set items:")
for item in numbers:
    print(item)
```

## 9. Program to Perform Set Operations

This program demonstrates set operations: union, intersection, difference, and symmetric difference.

```python
# Create two sets
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}
print("Set A:", set_a)
print("Set B:", set_b)

# Union
print("Union (A | B):", set_a | set_b)

# Intersection
print("Intersection (A & B):", set_a & set_b)

# Difference
print("Difference (A - B):", set_a - set_b)
print("Difference (B - A):", set_b - set_a)

# Symmetric Difference
print("Symmetric Difference (A ^ B):", set_a ^ set_b)
```

## 10. Program to Use String Manipulation Methods and Perform String Operations

This program demonstrates string manipulation methods (e.g., upper, lower, strip, replace) and operations (e.g., concatenation, slicing, splitting).

```python
# Create a string
text = "  Hello, Python Programming!  "
print("Original string:", text)

# String manipulation methods
print("Uppercase:", text.upper())
print("Lowercase:", text.lower())
print("Stripped:", text.strip())
print("Replace 'Python' with 'World':", text.replace("Python", "World"))
print("Capitalized:", text.strip().capitalize())

# String operations
# Concatenation
greeting = "Welcome to " + text.strip()
print("Concatenation:", greeting)

# Slicing
print("Slice [7:13]:", text[7:13])  # Extracts 'Python'

# Splitting
words = text.strip().split()
print("Split into words:", words)

# Check if string starts/ends with substring
print("Starts with '  Hello':", text.startswith("  Hello"))
print("Ends with 'ing!  ':", text.endswith("ing!  "))
```
