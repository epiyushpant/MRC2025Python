## Section 1: Variables, Constants, Data Types, and Type Conversion

In this section, we explore:
- **Variables**: Containers for storing data values.
- **Constants**: Values that should not change during program execution (by convention in uppercase).
- **Data Types**: Common types like integers, floats, strings, booleans, lists, tuples, and dictionaries.
- **Type Conversion**: Converting between data types using functions like `int()`, `float()`, `str()`, etc.

```python
# Demonstrating variables, constants, data types, and type conversion

# Constants (by convention, use uppercase for constants)
PI = 3.14159
GRAVITY = 9.81

# Variables with different data types
integer_var = 42                # Integer
float_var = 3.14               # Float
string_var = "Hello, Python!"  # String
boolean_var = True             # Boolean
list_var = [1, 2, 3, 4, 5]    # List
tuple_var = (10, 20, 30)       # Tuple
dict_var = {"name": "Gopi", "age": 25}  # Dictionary

# Displaying variable values and their types
print("Variable Values and Types:")
print(f"integer_var: {integer_var}, Type: {type(integer_var)}")
print(f"float_var: {float_var}, Type: {type(float_var)}")
print(f"string_var: {string_var}, Type: {type(string_var)}")
print(f"boolean_var: {boolean_var}, Type: {type(boolean_var)}")
print(f"list_var: {list_var}, Type: {type(list_var)}")
print(f"tuple_var: {tuple_var}, Type: {type(tuple_var)}")
print(f"dict_var: {dict_var}, Type: {type(dict_var)}")
print(f"PI (constant): {PI}, Type: {type(PI)}")
print(f"GRAVITY (constant): {GRAVITY}, Type: {type(GRAVITY)}")

# Type conversion examples
print("\nType Conversion Examples:")
str_to_int = int("123")        # String to Integer
float_to_int = int(3.99)       # Float to Integer (truncates decimal)
int_to_float = float(42)       # Integer to Float
int_to_string = str(42)        # Integer to String
string_to_bool = bool("True")  # String to Boolean

print(f"String to Integer: '123' -> {str_to_int}, Type: {type(str_to_int)}")
print(f"Float to Integer: 3.99 -> {float_to_int}, Type: {type(float_to_int)}")
print(f"Integer to Float: 42 -> {int_to_float}, Type: {type(int_to_float)}")
print(f"Integer to String: 42 -> {int_to_string}, Type: {type(int_to_string)}")
print(f"String to Boolean: 'True' -> {string_to_bool}, Type: {type(string_to_bool)}")
```

## Section 2: Operators and Calculations

Python supports various operators:
- **Arithmetic**: `+`, `-`, `*`, `/`, `//`, `%`, `**`
- **Comparison**: `==`, `!=`, `>`, `<`, `>=`, `<=`
- **Logical**: `and`, `or`, `not`
- **Assignment**: `=`, `+=`, `-=`, `*=`, etc.
- **Bitwise**: `&`, `|`, `^`, `~`, `<<`, `>>`

The following code demonstrates these operators with sample calculations.

```python
# Demonstrating different types of operators in Python

# Sample numbers for calculations
a = 10
b = 3
c = 2.5

# Arithmetic Operators
print("Arithmetic Operators:")
print(f"{a} + {b} = {a + b}")        # Addition
print(f"{a} - {b} = {a - b}")        # Subtraction
print(f"{a} * {b} = {a * b}")        # Multiplication
print(f"{a} / {b} = {a / b}")        # Division
print(f"{a} // {b} = {a // b}")      # Floor Division
print(f"{a} % {b} = {a % b}")        # Modulus
print(f"{a} ** {b} = {a ** b}")      # Exponentiation

# Comparison Operators
print("\nComparison Operators:")
print(f"{a} == {b}: {a == b}")       # Equal
print(f"{a} != {b}: {a != b}")       # Not equal
print(f"{a} > {b}: {a > b}")         # Greater than
print(f"{a} < {b}: {a < b}")         # Less than
print(f"{a} >= {b}: {a >= b}")       # Greater than or equal
print(f"{a} <= {b}: {a <= b}")       # Less than or equal

# Logical Operators
x = True
y = False
print("\nLogical Operators:")
print(f"{x} and {y}: {x and y}")      # AND
print(f"{x} or {y}: {x or y}")        # OR
print(f"not {x}: {not x}")            # NOT

# Assignment Operators
print("\nAssignment Operators:")
num = 5
print(f"Initial num: {num}")
num += 3    # num = num + 3
print(f"After num += 3: {num}")
num *= 2    # num = num * 2
print(f"After num *= 2: {num}")

# Bitwise Operators
print("\nBitwise Operators:")
print(f"{a} & {b} = {a & b}")        # AND
print(f"{a} | {b} = {a | b}")        # OR
print(f"{a} ^ {b} = {a ^ b}")        # XOR
print(f"~{a} = {~a}")                # NOT
print(f"{a} << 1 = {a << 1}")        # Left Shift
print(f"{a} >> 1 = {a >> 1}")        # Right Shift

# Sample calculation using multiple operators
result = (a * b) + (c ** 2) - (a / b)
print(f"\nSample Calculation: ({a} * {b}) + ({c} ** 2) - ({a} / {b}) = {result}")
```

## Section 3: Input/Output Functions

Python provides functions for input and output:
- **Input**: `input()` function to get user input as strings.
- **Output**: `print()` function with various formatting options (f-strings, `.format()`, `%` operator).

The following code demonstrates interactive I/O with a simple calculator example, including error handling to prevent crashes from invalid inputs.

```python
# Demonstrating Input/Output functions in Python with error handling

# Basic output using print()
print("Welcome to the I/O Demonstration Program!")
print("======================================")

# Getting user input with error handling
try:
    name = input("Please enter your name: ").strip()
    if not name:
        raise ValueError("Name cannot be empty.")
    age = int(input("Please enter your age: "))
    if age < 0:
        raise ValueError("Age cannot be negative.")

    # Formatted output using f-strings
    print(f"\nHello, {name}!")
    print(f"You are {age} years old.")

    # Multiple ways to format output
    print("\nDifferent ways to format output:")
    print(f"1. Using f-string: Name = {name}, Age = {age}")
    print("3. Using .format(): Name = {}, Age = {}".format(name, age))


    # Writing to a string (simulating file output in a browser-compatible way)
    output_text = f"User Profile:\nName: {name}\nAge: {age}\n"
    print("\nSimulated file output:")
    print(output_text)

except ValueError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"Unexpected error: {e}")

# Reading input and performing a simple calculation with error handling
print("\nLet's do a simple calculation!")
try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operation = input("Enter operation (+, -, *, /): ").strip()

    # Performing calculation based on input
    if operation == '+':
        result = num1 + num2
    elif operation == '-':
        result = num1 - num2
    elif operation == '*':
        result = num1 * num2
    elif operation == '/':
        if num2 == 0:
            raise ValueError("Division by zero is not allowed.")
        result = num1 / num2
    else:
        raise ValueError("Invalid operation. Please use +, -, *, or /.")

    # Outputting the result
    print(f"\nResult: {num1} {operation} {num2} = {result}")

except ValueError as e:
    print(f"Error: {e}")
except Exception as e:
    print(f"Unexpected error: {e}")
```
