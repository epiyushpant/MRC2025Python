## 1. Program to Divide Work in Functions

This program demonstrates modular programming by dividing a simple student grade calculator into functions: one to collect input, one to calculate the average, and one to determine the grade.

```python
def get_student_scores():
    """Collect scores for three subjects from the user."""
    scores = []
    for i in range(1, 4):
        score = float(input(f"Enter score for subject {i}: "))
        scores.append(score)
    return scores

def calculate_average(scores):
    """Calculate the average of a list of scores."""
    return sum(scores) / len(scores)

def determine_grade(average):
    """Determine grade based on average score."""
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

# Main program
scores = get_student_scores()
avg = calculate_average(scores)
grade = determine_grade(avg)
print(f"Scores: {scores}")
print(f"Average: {avg:.2f}")
print(f"Grade: {grade}")
```

## 2. Program with Different Types of Functions

This program demonstrates three types of functions:
- **Function with arguments**: Calculates the area of a rectangle.
- **Value-returning function**: Returns the square of a number.
- **Function without arguments**: Prints a greeting message.

```python
# Function with arguments
def calculate_area(length, width):
    """Calculate area of a rectangle given length and width."""
    return length * width

# Value-returning function
def square_number(num):
    """Return the square of a number."""
    return num * num

# Function without arguments
def greet():
    """Print a greeting message."""
    print("Hello, welcome to Python programming!")

# Test the functions
print("Area of rectangle (4, 5):", calculate_area(4, 5))
print("Square of 6:", square_number(6))
greet()
```

## 3. Program to Store Output in File

This program generates a multiplication table (1 to 10) and writes it to a file named `multiplication_table.txt`.

```python
# Generate multiplication table and write to file
try:
    with open('multiplication_table.txt', 'w') as file:
        for i in range(1, 11):
            for j in range(1, 11):
                result = i * j
                file.write(f"{i} x {j} = {result}\n")
        print("Multiplication table written to multiplication_table.txt")
except IOError:
    print("Error writing to file.")
```

## 4. Program to Read Input from File

This program reads the multiplication table from `multiplication_table.txt` (created above) and prints its contents. Note: Ensure the file exists before running this code.

```python
# Read from file
try:
    with open('multiplication_table.txt', 'r') as file:
        content = file.readlines()
        print("Contents of multiplication_table.txt:")
        for line in content:
            print(line.strip())
except FileNotFoundError:
    print("File not found.")
except IOError:
    print("Error reading file.")
```

## 5. Program to Handle Different Types of Exceptions

This program demonstrates handling various exceptions:
- `ValueError`: For invalid input (e.g., non-numeric input).
- `ZeroDivisionError`: For division by zero.
- `IndexError`: For accessing an invalid list index.
- `FileNotFoundError`: For missing files (already shown above).

```python
# Program to demonstrate exception handling
def perform_operations():
    try:
        # Input for division
        num1 = float(input("Enter a number: "))
        num2 = float(input("Enter another number to divide by: "))
        result = num1 / num2
        print(f"Division result: {result}")

        # List access
        my_list = [1, 2, 3]
        index = int(input("Enter an index to access list (0-2): "))
        print(f"Element at index {index}: {my_list[index]}")

    except ValueError:
        print("Error: Please enter valid numbers.")
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
    except IndexError:
        print("Error: Invalid index. Please use 0, 1, or 2.")
    except Exception as e:
        print(f"Unexpected error: {e}")
    finally:
        print("Operation completed (success or failure).")

# Run the program
perform_operations()
```
