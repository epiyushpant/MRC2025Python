```python
import tkinter as tk

root = tk.Tk()            # create the window
root.title("My App")      # set the title bar text
root.geometry("400x300")  # set size: width x height

root.mainloop()           # keep the window open
```

```python
import tkinter as tk

root = tk.Tk()
root.title("Button Example")
root.geometry("300x200")

# This function runs when button is clicked
def say_hello():
    print("Hello! Button was clicked!")

# Create the button
btn = tk.Button(root, text="Click Me!", command=say_hello)
btn.pack(pady=30)

root.mainloop()
```
