```python
# =============================================
# COMPREHENSIVE PYTHON OPERATORS DEMONSTRATION
# =============================================

def demonstrate_operators():
    print("=" * 60)
    print("PYTHON OPERATORS DEMONSTRATION")
    print("=" * 60)
    
    # Sample values for demonstration
    a, b, c = 15, 4, 3
    x, y = 7.5, 2.5
    str1, str2 = "Hello", "World"
    list1, list2 = [1, 2, 3], [4, 5, 6]
    
    # 1. ARITHMETIC OPERATORS
    print("\n1. ARITHMETIC OPERATORS")
    print("=" * 30)
    print(f"{a} + {b} = {a + b}")      # Addition
    print(f"{a} - {b} = {a - b}")      # Subtraction
    print(f"{a} * {b} = {a * b}")      # Multiplication
    print(f"{a} / {b} = {a / b}")      # Division
    print(f"{a} // {b} = {a // b}")    # Floor Division
    print(f"{a} % {b} = {a % b}")      # Modulus
    print(f"{a} ** {c} = {a ** c}")    # Exponentiation
    
    # 2. COMPARISON OPERATORS
    print("\n2. COMPARISON OPERATORS")
    print("=" * 30)
    print(f"{a} == {b} : {a == b}")    # Equal to
    print(f"{a} != {b} : {a != b}")    # Not equal to
    print(f"{a} > {b}  : {a > b}")     # Greater than
    print(f"{a} < {b}  : {a < b}")     # Less than
    print(f"{a} >= {b} : {a >= b}")    # Greater than or equal to
    print(f"{a} <= {b} : {a <= b}")    # Less than or equal to
    
    # 3. LOGICAL OPERATORS
    print("\n3. LOGICAL OPERATORS")
    print("=" * 30)
    print(f"({a} > 10) and ({b} < 5) : {(a > 10) and (b < 5)}")   # AND
    print(f"({a} > 20) or ({b} < 5)  : {(a > 20) or (b < 5)}")    # OR
    print(f"not ({a} > 10)           : {not (a > 10)}")           # NOT
    
    # 4. ASSIGNMENT OPERATORS
    print("\n4. ASSIGNMENT OPERATORS")
    print("=" * 30)
    num = a
    print(f"num = {a} → {num}")
    
    num += b  # num = num + b
    print(f"num += {b} → {num}")
    
    num -= c  # num = num - c
    print(f"num -= {c} → {num}")
    
    num *= 2  # num = num * 2
    print(f"num *= 2 → {num}")
    
    num /= 2  # num = num / 2
    print(f"num /= 2 → {num}")
    
    # 5. BITWISE OPERATORS
    print("\n5. BITWISE OPERATORS")
    print("=" * 30)
    print(f"{a} & {b}  = {a & b}")     # AND
    print(f"{a} | {b}  = {a | b}")     # OR
    print(f"{a} ^ {b}  = {a ^ b}")     # XOR
    print(f"~{a}       = {~a}")        # NOT
    print(f"{a} << 1   = {a << 1}")    # Left shift
    print(f"{a} >> 1   = {a >> 1}")    # Right shift
    
    # 6. IDENTITY OPERATORS
    print("\n6. IDENTITY OPERATORS")
    print("=" * 30)
    list3 = list1
    print(f"list1 is list3     : {list1 is list3}")
    print(f"list1 is not list2 : {list1 is not list2}")
    
    # 7. MEMBERSHIP OPERATORS
    print("\n7. MEMBERSHIP OPERATORS")
    print("=" * 30)
    print(f"{c} in {list1}     : {c in list1}")
    print(f"{b} not in {list1} : {b not in list1}")
    print(f"'e' in '{str1}'    : {'e' in str1}")
    
   
# Run the demonstration
if __name__ == "__main__":
    demonstrate_operators()
    
    # Additional quick calculations
    print("\n" + "=" * 60)
    print("QUICK CALCULATIONS")
    print("=" * 60)
    
    # Quick math operations
    calculations = [
        ("2 + 3 * 4", 2 + 3 * 4),
        ("(2 + 3) * 4", (2 + 3) * 4),
        ("10 / 3", 10 / 3),
        ("10 // 3", 10 // 3),
        ("2 ** 8", 2 ** 8),
        ("17 % 5", 17 % 5),
        ("5 & 3", 5 & 3),
        ("5 | 3", 5 | 3),
        ("~5", ~5)
    ]
    
    for expr, result in calculations:
        print(f"{expr:15} = {result}")
```

```python
# =============================================
# PYTHON INPUT/OUTPUT FUNCTIONS DEMONSTRATION
# =============================================

def demonstrate_io():
    print("=" * 50)
    print("PYTHON I/O FUNCTIONS")
    print("=" * 50)
    

    # 1. FILE I/O OPERATIONS
    print("\n1. FILE I/O OPERATIONS")
    print("-" * 30)
    
    # Writing to a file
    with open('demo_file.txt', 'w') as file:
        file.write("Hello, this is line 1\n")
        file.write("This is line 2\n")
    print("File written successfully!")
    
    # Reading from a file
    print("\nReading file contents:")
    with open('demo_file.txt', 'r') as file:
        content = file.read()
        print(content)
    
    # Reading line by line
    print("Reading line by line:")
    with open('demo_file.txt', 'r') as file:
        for i, line in enumerate(file, 1):
            print(f"Line {i}: {line.strip()}")
    
    
 

# Run the demonstration
if __name__ == "__main__":
    demonstrate_io()
    
    # Clean up created file
    import os
    if os.path.exists('demo_file.txt'):
        os.remove('demo_file.txt')
        print("\nTemporary file cleaned up!")
```

```python
# Simple program to demonstrate if, match, break and continue

while True:
    # Take user input
    number = int(input("Enter a number (0 to quit): "))

    # if statement
    if number == 0:
        print("You entered 0. Exiting...")
        break  # break out of the loop

    # if-else decision
    if number > 0:
        print("The number is positive.")
    else:
        print("The number is negative.")

    # match statement (Python 3.10+)
    match number % 2:
        case 0:
            print("It is an even number.")
        case 1:
            print("It is an odd number.")
        case _:
            print("Unknown case")

    # continue example
    # If the number is negative, skip the rest of the loop
    if number < 0:
        print("Skipping the rest of the loop for negative number.\n")
        continue

    print("This line runs only for positive numbers.\n")
```

```python
# Program demonstrating different loop statements to solve meaningful problems

# 1. Using a FOR loop to print a multiplication table
print("----- Multiplication Table of 5 -----")
for i in range(1, 11):
    print(f"5 x {i} = {5 * i}")

print("\n")

# 2. Using a WHILE loop to calculate the sum of user-entered numbers
total = 0
print("----- Sum of User Entered Numbers (enter 0 to stop) -----")
while True:
    num = int(input("Enter a number (0 to stop): "))
    if num == 0:
        break  # Exit the loop when user enters 0
    total += num

print(f"Total sum = {total}")

print("\n")

# 3. Using NESTED loops to print a pattern (right-angled triangle of stars)
print("----- Star Pattern -----")
rows = 5
for i in range(1, rows + 1):
    for j in range(i):
        print("*", end="")
    print()  # Moves to the next line
```

```python
# Program to demonstrate input validation using a loop

while True:
    user_input = input("Enter a number between 1 and 10: ")

    # Check if the input is a digit (only numbers allowed)
    if user_input.isdigit():
        number = int(user_input)

        # Check if number is in valid range
        if 1 <= number <= 10:
            print("Valid input! You entered:", number)
            break  # Exit the loop if input is valid
        else:
            print("Error: Number must be between 1 and 10. Please try again.")

    else:
        print("Error: Invalid input. Please enter a number (digits only).")
```

```python
# Program to create different patterns using nested loops

# 1. Right-angled Triangle
print("Pattern 1: Right-angled Triangle")
rows = 5
for i in range(1, rows + 1):
    for j in range(1, i + 1):
        print("*", end="")
    print()
print()

# 2. Inverted Right-angled Triangle
print("Pattern 2: Inverted Right-angled Triangle")
for i in range(rows, 0, -1):
    for j in range(1, i + 1):
        print("*", end="")
    print()
print()

# 3. Pyramid Pattern
print("Pattern 3: Pyramid")
for i in range(1, rows + 1):
    # Print spaces
    for j in range(rows - i):
        print(" ", end="")
    # Print stars
    for k in range(2 * i - 1):
        print("*", end="")
    print()
print()

# 4. Diamond Pattern
print("Pattern 4: Diamond")
# Upper half
for i in range(1, rows + 1):
    for j in range(rows - i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("*", end="")
    print()
# Lower half
for i in range(rows - 1, 0, -1):
    for j in range(rows - i):
        print(" ", end="")
    for k in range(2 * i - 1):
        print("*", end="")
    print()
```

```python
# Program demonstrating the use of an infinite loop

while True:
    print("----- Menu -----")
    print("1. Say Hello")
    print("2. Say Goodbye")
    print("3. Exit")

    choice = input("Enter your choice (1/2/3): ")

    if choice == "1":
        print("Hello!\n")
    elif choice == "2":
        print("Goodbye!\n")
    elif choice == "3":
        print("Exiting the program...")
        break   # break stops the infinite loop
    else:
        print("Invalid choice. Please try again.\n")
```

```python
# Program to create a list, add elements, remove elements and display list items

# Step 1: Create an empty list
my_list = []

# Step 2: Add elements to the list
my_list.append("Apple")
my_list.append("Banana")
my_list.append("Orange")

# Step 3: Display the list after adding elements
print("List after adding elements:", my_list)

# Step 4: Remove an element from the list
my_list.remove("Banana")

# Step 5: Display the list after removing an element
print("List after removing an element:", my_list)

# Step 6: Display each item in the list
print("Displaying all list items:")
for item in my_list:
    print(item)
```

```python
# Program to demonstrate list slicing

# Creating a list of numbers
numbers = [10, 20, 30, 40, 50, 60, 70, 80]

# Display the full list
print("Full list:", numbers)

# Display the first 4 elements
print("First 4 elements:", numbers[0:4])  

# Display elements from index 2 to 5
print("Elements from index 2 to index 5:", numbers[2:6])

# Display all elements from index 3 to the end
print("Elements from index 3 to end:", numbers[3:])

# Display every second element
print("Every second element:", numbers[::2])
```

```python
# Program to demonstrate list comprehension

# 1. Create a list of squares of numbers from 1 to 10
squares = [x*x for x in range(1, 11)]
print("Squares:", squares)

# 2. Create a list of even numbers from 1 to 20
evens = [x for x in range(1, 21) if x % 2 == 0]
print("Even numbers:", evens)

# 3. Convert all items in a list to uppercase using list comprehension
fruits = ["apple", "banana", "orange", "mango"]
fruits_upper = [fruit.upper() for fruit in fruits]
print("Fruits in uppercase:", fruits_upper)
```
