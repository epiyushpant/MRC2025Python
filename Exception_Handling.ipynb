<h1>Exception Handling in Python</h1>

<h2>Basic Syntax of Exception Handling</h2>

```python
try:
    # Code that might raise an exception
    pass
except ExceptionType:
    # Handle the exception
    pass
else:
    # Executes if no exception occurs
    pass
finally:
    # Always executes, regardless of exception
    pass
```

<h2>Example 1 - Handling ZeroDivisionError</h2>

```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ZeroDivisionError:
    print("Error: Division by zero is not allowed!")
else:
    print(f"Result: {result}")
finally:
    print("Execution complete.")
```

```python
num = int(input("Enter a number: "))
result = 10 / num
print(result)
```

<p>Explanation:<br>
    <li>Attempts to divide 10 by user input.</li>
    <li>Catches division by zero error.</li>
    <li>else runs if no error; finally always runs.</li>

<h2>Example 2 - Handling FileNotFoundError</h2>

```python
try:
    with open("data.txt", "r") as file:
        content = file.read()
except FileNotFoundError:
    print("Error: The file 'data.txt' was not found!")
else:
    print("File contents:", content)
finally:
    print("File operation complete.")
```

<p>Explanation:<br>
<li>Tries to read a file.</li>
<li>Handles case where file doesn't exist.</li>
<li>Uses with for proper file handling.</li>

<h2>Example 3 - Multiple Exception Handling</h2>

```python
try:
    numbers = [1, 2, 3]
    index = int(input("Enter index: "))
    value = numbers[index]
except IndexError:
    print("Error: Index out of range!")
except ValueError:
    print("Error: Please enter a valid integer!")
else:
    print(f"Value at index {index}: {value}")
finally:
    print("Operation complete.")
```

<p>Explanation:<br>
    <li>Handles multiple exceptions (IndexError, ValueError).</li>
    <li>Demonstrates catching specific exceptions separately.</li>

<h2>Example 4 - Custom Exceptions</h2>

```python
class InvalidBalanceError(Exception):
    """Custom exception for invalid balance."""
    pass

class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        if amount > self.balance:
            raise InvalidBalanceError("Withdrawal amount exceeds balance!")
        self.balance -= amount
        return self.balance

try:
    account = BankAccount(100)
    new_balance = account.withdraw(150)
except InvalidBalanceError as e:
    print(f"Error: {e}")
else:
    print(f"New balance: {new_balance}")
finally:
    print("Transaction complete.")
```

<p>Explanation:<br>
    <li>Create custom exceptions by subclassing Exception.</li>
    <li>Defines InvalidBalanceError for specific error cases.</li>
    <li>Raises custom exception in BankAccount class.</li>
    <li>Handles custom exception with meaningful feedback.</li>

<h2>Raising Exceptions</h2>

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older!")
    return f"Age {age} is valid."

try:
    result = check_age(15)
except ValueError as e:
    print(f"Error: {e}")
else:
    print(result)
```

<p>Explanation:<br>
    <li>Use raise to trigger exceptions manually.</li>
    <li>Custom validation with raise.</li>
    <li>Captures and displays the error message.</li>
