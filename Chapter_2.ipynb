<p>1.Write program to apply if, match, break and continue statements for decision making.</p>

```python
# Using if, match, break, and continue
while True:
    number = int(input("Enter a number (0 to exit): "))
    
    if number == 0:
        print("Exiting program...")
        break  # exits the loop
    
    if number < 0:
        print("Negative numbers not allowed. Try again.")
        continue  # skips the rest of the loop

    match number % 2:
        case 0:
            print("Even number")
        case 1:
            print("Odd number")
```

2.Write program to utilize different loop statements to solve meaningful problems.

```python
# Print the sum of first N natural numbers using different loops

n = int(input("Enter a number: "))

# Using for loop
sum1 = 0
for i in range(1, n + 1):
    sum1 += i
print(f"Sum using for loop: {sum1}")

# Using while loop
sum2 = 0
i = 1
while i <= n:
    sum2 += i
    i += 1
print(f"Sum using while loop: {sum2}")
```

3.Write program to demonstrate input validation using loop.

```python
# Validate age input between 1 and 100
while True:
    age = input("Enter your age (1-100): ")
    if age.isdigit() and 1 <= int(age) <= 100:
        print("Valid age entered!")
        break
    else:
        print("Invalid input. Please enter a number between 1 and 100.")
```

4.Write program to create different patterns using nested loop.

```python
# Pattern: Right-angled triangle
rows = int(input("Enter number of rows: "))

for i in range(1, rows + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```
